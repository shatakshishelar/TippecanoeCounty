```r
library(stringr)
library(dplyr)
library(ggplot2)
library(data.table)
library(readxl)
```

```r
combinedwkbk23 = read.csv('tippecanoe/Combined_Workbook_2023.csv')
combinedwkbk24 = read.csv('tippecanoe/Combined_Workbook_2024.csv')
combinedexc23 = read.csv('tippecanoe/Combined_Reconciliation_2023.csv')
combinedexc24 = read.csv('tippecanoe/Combined_Reconciliation_2024.csv')
combinedrat23 = read.csv('tippecanoe/Combined_Study_2023.csv')
combinedrat24 = read.csv('tippecanoe/Combined_Study_2024.csv')
```

```r
## Prints out plots of the Top 20 Exclusion Reasons for each county in 2024
for (county in unique(combinedexc24$County)) {
    excludedbook = combinedexc24[combinedexc24$County == county,]
    top_20_explanations <- as.data.frame(table(excludedbook$Explanation))
    colnames(top_20_explanations) <- c("Explanation", "Frequency")
    top_20_explanations <- top_20_explanations[order(-top_20_explanations$Frequency), ][1:20, ]
    if (nrow(top_20_explanations) > 0) {
        print(
            ggplot(top_20_explanations, aes(x = reorder(Explanation, Frequency), y = Frequency)) +
            geom_bar(stat = "identity", fill = "lightblue") +
            coord_flip() +
            theme_minimal() +
            labs(
                title = paste("Top 20 Explanations by Frequency -", county),
                x = "Explanation",
                y = "Frequency"
            ) +
            theme(axis.text.x = element_text(angle = 70, hjust = 1))
            + scale_x_discrete(labels = function(x) str_trunc(x, width = 35, side = "right"))
        )
    } else {
        message("No valid data in 'Explanation' column for county: ", county)
    }
}
```

```r
## Prints out the Top 40 Statewide exclusion reasons in 2024
top_40_explanations <- as.data.frame(table(combinedexc24$Explanation[combinedexc24$Reclassified]))
    colnames(top_40_explanations) <- c("Explanation", "Frequency")
    
    # Filter for top 30 explanations
    top_40_explanations <- top_40_explanations[order(-top_40_explanations$Frequency), ][1:40, ]
    
    # Plotting in Jupyter Notebook
    if (nrow(top_40_explanations) > 0) {
        print(
            ggplot(top_40_explanations, aes(x = reorder(Explanation, Frequency), y = Frequency)) +
            geom_bar(stat = "identity", fill = "lightblue") +
            coord_flip() +
            theme_minimal() +
            labs(
                title = paste("Top 40 Explanations by Frequency Total"),
                x = "Explanation",
                y = "Frequency"
            ) +
            theme(axis.text.x = element_text(angle = 70, hjust = 1))
        )
    } else {
        message("No valid data in 'Explanation' column for county: ", county)
    }
```

```r
## Prints out the Top 40 Statewide exclusion reasons in 2023
top_40_explanations <- as.data.frame(table(combinedexc23$Explanation[combinedexc23$Reclassified]))
    colnames(top_40_explanations) <- c("Explanation", "Frequency")
    top_40_explanations <- top_40_explanations[order(-top_40_explanations$Frequency), ][1:40, ]
    if (nrow(top_40_explanations) > 0) {
        print(
            ggplot(top_40_explanations, aes(x = reorder(Explanation, Frequency), y = Frequency)) +
            geom_bar(stat = "identity", fill = "lightblue") +
            coord_flip() +
            theme_minimal() +
            labs(
                title = paste("Top 40 Explanations by Frequency Total"),
                x = "Explanation",
                y = "Frequency"
            ) +
            theme(axis.text.x = element_text(angle = 70, hjust = 1))
        )
    } else {
        message("No valid data in 'Explanation' column for county: ", county)
    }
```

```r
## 2023 Top 40 Exclusion Reasons
## Colored by inclusion in aggressive ratio study only, aggressive & moderate ratio studies, or none
aggressive = c('AV DOES NOT REFLECT MARKET VALUE','MULTI PARCEL SALE','INVESTMENT/RENTAL','SOLD AS IS',
               'INSUFFICIENT TIME ON MARKET','VACANT LAND','APARTMENT','LEASE','BUILDING PERMIT',
               'NOT ENOUGH SALES FOR TRENDING')
moderate = c('AV DOES NOT REFLECT MARKET VALUE','MULTI PARCEL SALE','SOLD AS IS','INSUFFICIENT TIME ON MARKET',
              'VACANT LAND','BUILDING PERMIT','NOT ENOUGH SALES FOR TRENDING')

top_40_explanations$FillColor <- case_when(
  top_40_explanations$Explanation %in% aggressive & top_40_explanations$Explanation %in% moderate ~ "green",
  top_40_explanations$Explanation %in% aggressive & !(top_40_explanations$Explanation %in% moderate) ~ "yellow",
  TRUE ~ "red"
)

top_40_explanations$Category <- case_when(
  top_40_explanations$FillColor == "green" ~ "Definite Inclusion (2)",
  top_40_explanations$FillColor == "yellow" ~ "Possible Inclusion (1)",
  top_40_explanations$FillColor == "red" ~ "Definite Exclusion (0 & NA)"
)

if (nrow(top_40_explanations) > 0) {
    print(
        ggplot(top_40_explanations, aes(x = reorder(Explanation, Frequency), y = Frequency, fill = Category)) +
        geom_bar(stat = "identity") +
        scale_fill_manual(
            values = c("Definite Inclusion (2)" = "green",
                       "Possible Inclusion (1)" = "yellow",
                       "Definite Exclusion (0 & NA)" = "red"),
            name = "Explanation Evaluation"
        ) +
        coord_flip() +
        theme_minimal() +
        labs(
            title = paste("Top 40 Explanations by Frequency Total"),
            x = "Explanation",
            y = "Frequency"
        ) +
        theme(axis.text.x = element_text(angle = 70, hjust = 1))
    )
} else {
    message("No valid data in 'Explanation' column for county: ", county)
}
```
